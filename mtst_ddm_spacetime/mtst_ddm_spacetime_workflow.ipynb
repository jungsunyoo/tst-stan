# MTST DDM: space (memory) – time tradeoff

This notebook fits a targeted DDM model set for the hypothesis that conjunctive menu caching reflects a **memory–time tradeoff**. The core prediction is that cache / interference / menu-space variables should load primarily onto **nondecision time** (`t0`), not just drift (`v`).


```python
from pathlib import Path
DATA_DIR = Path(".")
print(DATA_DIR.resolve())
```

## Recommended fitting sequence

1. `baseline`
2. `t0_time`
3. `t0_space_time`  *(main model)*
4. `v_space_time`
5. `t0v_space_time`


```python
# Example shell commands (run in terminal or via ! if desired)
cmds = [
    "python fit_ddm_spacetime_hier.py --data_dir . --states 3 4 5 --variant baseline --K 1 --warmup 500 --draws 500 --outdir ddm_spacetime_baseline",
    "python fit_ddm_spacetime_hier.py --data_dir . --states 3 4 5 --variant t0_time --K 1 --warmup 500 --draws 500 --outdir ddm_spacetime_t0time",
    "python fit_ddm_spacetime_hier.py --data_dir . --states 3 4 5 --variant t0_space_time --K 1 --warmup 500 --draws 500 --outdir ddm_spacetime_t0spacetime",
    "python fit_ddm_spacetime_hier.py --data_dir . --states 3 4 5 --variant v_space_time --K 1 --warmup 500 --draws 500 --outdir ddm_spacetime_vspacetime",
    "python fit_ddm_spacetime_hier.py --data_dir . --states 3 4 5 --variant t0v_space_time --K 1 --warmup 500 --draws 500 --outdir ddm_spacetime_t0vspacetime",
]
for c in cmds:
    print(c)
```

## What to look for

The space–time tradeoff is strongest if `t0_space_time` wins or nearly wins model comparison, and if the `b_t0` coefficients show:
- `full_hit < 0`
- `neighbor_hit < 0`
- `log1p_menu_lag > 0`
- `diversity_prop > 0`
- `diversity_x_logM > 0`
- `full_hit_x_logM < 0`
